```python
import pandas as pd
data=pd.read_csv("student_exam_performance.csv")
data
```

```python
cat_cols = ['parent_education', 'notes_quality', 'sleep_quality', 'device_availability']

# Imputation par 'Inconnu'
data[cat_cols] = data[cat_cols].fillna('Inconnu')
```

```python
data["device_availability"].value_counts()
```

```python

num_cols = ['previous_gpa', 'attendance_percentage', 'time_management_score']
for col in num_cols:
    data[col] = data[col].fillna(data[col].median())
```
